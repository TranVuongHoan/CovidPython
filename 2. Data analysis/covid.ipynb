```python
import seaborn as sns

import matplotlib.pyplot as plt

# 1. Epidemic Curve by Country (Confirmed Cases Over Time)
def plot_country_curve(country, df):
    df_c = df[df["Country/Region"] == country].copy()
    df_c["Date"] = pd.to_datetime(df_c["Date"])
    plt.figure(figsize=(10,5))
    plt.plot(df_c["Date"], df_c["Confirmed"], label="Confirmed")
    plt.plot(df_c["Date"], df_c["Deaths"], label="Deaths")
    plt.plot(df_c["Date"], df_c["Recovered"], label="Recovered")
    plt.title(f"COVID-19 Curve in {country}")
    plt.xlabel("Date")
    plt.ylabel("Count")
    plt.legend()
    plt.show()

# 2. When Did the Outbreak Start to Surge in Each Country? (Detect Surge Start)
def detect_surge_start(country, df, threshold=1.5):
    df_c = df[df["Country/Region"] == country].copy()
    df_c["Date"] = pd.to_datetime(df_c["Date"])
    rolling_mean = df_c["New cases"].rolling(window=7).mean()
    surge = df_c[df_c["New cases"] > threshold * rolling_mean]
    if not surge.empty:
        print(f"Surge starts in {country} on:", surge.iloc[0]["Date"])
    else:
        print(f"No surge detected for {country}.")

# 3. Compare the Growth Rate of Confirmed Cases Among Countries in the Same Region
def compare_growth(region, df):
    countries = df[df["WHO Region"] == region]["Country/Region"].unique()
    plt.figure(figsize=(12,6))
    for country in countries:
        df_c = df[df["Country/Region"] == country].copy()
        df_c["Date"] = pd.to_datetime(df_c["Date"])
        plt.plot(df_c["Date"], df_c["Confirmed"], label=country)
    plt.title(f"Growth Comparison in {region}")
    plt.xlabel("Date")
    plt.ylabel("Confirmed Cases")
    plt.legend()
    plt.show()

# 4. Check for Seasonality (e.g., Case Increases During Winter)
def plot_monthly_cases(country, df):
    df_c = df[df["Country/Region"] == country].copy()
    df_c["Date"] = pd.to_datetime(df_c["Date"])
    df_c["Month"] = df_c["Date"].dt.month
    monthly_cases = df_c.groupby("Month")["New cases"].sum()
    plt.figure(figsize=(8,4))
    sns.barplot(x=monthly_cases.index, y=monthly_cases.values)
    plt.title(f"Monthly New Cases in {country}")
    plt.xlabel("Month")
    plt.ylabel("New Cases")
    plt.show()

# 5. How Did the Number of Deaths Change After Vaccine Rollout?
# Assume Vaccination Began in December 2020
def plot_deaths_after_vaccine(country, df):
    df_c = df[df["Country/Region"] == country].copy()
    df_c["Date"] = pd.to_datetime(df_c["Date"])
    vaccine_date = pd.to_datetime("2020-12-01")
    before = df_c[df_c["Date"] < vaccine_date]["Deaths"]
    after = df_c[df_c["Date"] >= vaccine_date]["Deaths"]
    plt.figure(figsize=(10,5))
    plt.plot(df_c["Date"], df_c["Deaths"], label="Deaths")
    plt.axvline(vaccine_date, color='red', linestyle='--', label='Vaccine rollout')
    plt.title(f"Deaths Before/After Vaccine in {country}")
    plt.xlabel("Date")
    plt.ylabel("Deaths")
    plt.legend()
    plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.neural_network import MLPRegressor
from scipy.signal import find_peaks
import pandas as pd

# 1. Predict the Number of Confirmed Cases for the Next 7 Days in a Specific Country (e.g., US)

# Prepare time series data for the selected country
country = "US"
df_country = dataset2[dataset2["Country/Region"] == country].copy()
df_country["Date"] = pd.to_datetime(df_country["Date"])
df_country = df_country.sort_values("Date")

# Use confirmed cases for forecasting
window = 7
X = []
y = []
for i in range(len(df_country) - window):
    X.append(df_country["Confirmed"].iloc[i:i+window].values)
    y.append(df_country["Confirmed"].iloc[i+window])
X = np.array(X)
y = np.array(y)

# Train/test split
split_idx = int(len(X) * 0.8)
X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]

# Train Random Forest model
rf_ts = RandomForestRegressor(random_state=42)
rf_ts.fit(X_train, y_train)

# Predict for the next 7 days
last_window = df_country["Confirmed"].iloc[-window:].values.reshape(1, -1)
future_preds = []
current_window = last_window.copy()
for _ in range(7):
    pred = rf_ts.predict(current_window)[0]
    future_preds.append(pred)
    current_window = np.roll(current_window, -1)
    current_window[0, -1] = pred

print("Predicted confirmed cases for the next 7 days in", country, ":", future_preds)

# 2. Predict Next-Day Deaths Using Random Forest, Gradient Boosting, and MLP

# Prepare death data
deaths = df_country["Deaths"].values
X_death = []
y_death = []
for i in range(len(deaths) - window):
    X_death.append(deaths[i:i+window])
    y_death.append(deaths[i+window])
X_death = np.array(X_death)
y_death = np.array(y_death)

X_train_d, X_test_d = X_death[:split_idx], X_death[split_idx:]
y_train_d, y_test_d = y_death[:split_idx], y_death[split_idx:]

# Random Forest
rf_death = RandomForestRegressor(random_state=42)
rf_death.fit(X_train_d, y_train_d)
rf_pred = rf_death.predict(X_test_d)

# Gradient Boosting
gb_death = GradientBoostingRegressor(random_state=42)
gb_death.fit(X_train_d, y_train_d)
gb_pred = gb_death.predict(X_test_d)

# MLP Regressor
mlp_death = MLPRegressor(random_state=42, max_iter=500)
mlp_death.fit(X_train_d, y_train_d)
mlp_pred = mlp_death.predict(X_test_d)

print("Random Forest MSE:", mean_squared_error(y_test_d, rf_pred))
print("Gradient Boosting MSE:", mean_squared_error(y_test_d, gb_pred))
print("MLP MSE:", mean_squared_error(y_test_d, mlp_pred))

# Predict next day's deaths
next_rf = rf_death.predict(deaths[-window:].reshape(1, -1))[0]
next_gb = gb_death.predict(deaths[-window:].reshape(1, -1))[0]
next_mlp = mlp_death.predict(deaths[-window:].reshape(1, -1))[0]
print("Predicted next-day deaths (RF, GB, MLP):", next_rf, next_gb, next_mlp)

# 3. Forecast When the Next Epidemic Peak May Occur (Simple Peak Detection)
peaks, _ = find_peaks(df_country["New cases"].values, distance=7)
peak_dates = df_country["Date"].iloc[peaks]
print("Dates of epidemic peaks:", peak_dates.tolist())

# 4. Classify “Surge Days” – Days When New Cases Increase >50% Compared to the Previous 7-Day Average
surge_threshold = 1.5
rolling_mean = df_country["New cases"].rolling(window=7).mean()
surge_days = df_country["Date"][(df_country["New cases"] > surge_threshold * rolling_mean)]
print("Detected surge days:", surge_days.tolist())
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.preprocessing import StandardScaler

# ===============================
# 1. Define columns to drop
# ===============================
drop_cols = ['Country/Region', 'Continent', 'WHO Region', 'iso_alpha', 'TotalDeaths']
drop_cols = [c for c in drop_cols if c in dataset1.columns]  # drop only if present

# ===============================
# 2. Clean dataset
# ===============================
# Drop rows where target is NaN
mask = dataset1['TotalDeaths'].notna()
X = dataset1.drop(columns=drop_cols, errors="ignore").loc[mask]
y = dataset1.loc[mask, 'TotalDeaths']

# Keep only numeric features
X = X.select_dtypes(include=[np.number])

# Fill missing values in features with median
X = X.fillna(X.median(numeric_only=True))

# ===============================
# 3. Feature scaling
# ===============================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ===============================
# 4. Train/test split
# ===============================
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# ===============================
# 5. Random Forest + GridSearchCV
# ===============================
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [5, 10, 15, None],
    'min_samples_split': [2, 5, 10]
}
rf = RandomForestRegressor(random_state=42)
grid_search = GridSearchCV(
    rf, param_grid, cv=5, scoring='neg_mean_squared_error', n_jobs=-1
)
grid_search.fit(X_train, y_train)

best_rf = grid_search.best_estimator_

# ===============================
# 6. Cross-validation
# ===============================
cv_scores = cross_val_score(best_rf, X_train, y_train, cv=5, scoring='neg_mean_squared_error')

# ===============================
# 7. Predictions & Evaluation
# ===============================
y_pred_rf = best_rf.predict(X_test)

mse = mean_squared_error(y_test, y_pred_rf)
mae = mean_absolute_error(y_test, y_pred_rf)
r2 = r2_score(y_test, y_pred_rf)

print("Best Random Forest Parameters:", grid_search.best_params_)
print("Cross-validated MSE:", -np.mean(cv_scores))
print("Test MSE:", mse)
print("Test MAE:", mae)
print("Test R2:", r2)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, roc_curve, auc
import seaborn as sns

# Ensure Date column is in datetime format
dataset2['Date'] = pd.to_datetime(dataset2['Date'], errors='coerce')

# ========== 1. Cluster Countries by Epidemic Curve Shape ==========
pivot = dataset2.pivot_table(index='Country/Region', columns='Date', values='Confirmed', fill_value=0)
X_curve = StandardScaler().fit_transform(pivot)

# Dimensionality reduction using PCA
pca = PCA(n_components=10)
X_pca = pca.fit_transform(X_curve)

# KMeans clustering
n_clusters = 3
kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
labels = kmeans.fit_predict(X_pca)

# Visualize clusters
plt.figure(figsize=(8,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=labels, cmap='viridis')
plt.title('Country Clusters by Epidemic Curve Shape')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.show()

# ========== 2. Classification: "Good Control" vs "Poor Control" ==========
surge_threshold = 1.5  # Example: a surge occurs when new cases > 150% of the 7-day average

# Select data from the most recent 60 days
recent = dataset2[dataset2['Date'] > (dataset2['Date'].max() - pd.Timedelta(days=60))]

# Count number of surge days in the last 60 days
surge_counts = recent.groupby('Country/Region')['New cases'].apply(
    lambda x: (x > surge_threshold * x.rolling(7, min_periods=1).mean()).sum()
)

# Label countries: 1 = good control (few surges), 0 = poor control
control_label = (surge_counts < 2).astype(int)

# Prepare features from dataset1
features = dataset1.set_index('Country/Region')[['TotalCases','TotalDeaths','TotalTests','Population']].fillna(0)
X_class = StandardScaler().fit_transform(features)
y_class = control_label.reindex(features.index).fillna(0).astype(int)

# Train classifier
clf = RandomForestClassifier(random_state=42)
clf.fit(X_class, y_class)
print("Classification accuracy:", clf.score(X_class, y_class))

# ========== 3. Predict Probability of a Surge in the Next 14 Days ==========
def make_surge_dataset(df, window=30, pred_days=14, surge_threshold=1.5):
    X, y = [], []
    countries = df['Country/Region'].unique()
    for country in countries:
        df_c = df[df['Country/Region'] == country].sort_values('Date')
        cases = df_c['New cases'].fillna(0).values
        for i in range(len(cases) - window - pred_days):
            past_window = cases[i:i+window]
            future = cases[i+window:i+window+pred_days]
            ref_mean = pd.Series(past_window).rolling(7, min_periods=1).mean().iloc[-1]
            surge_flag = int((future > surge_threshold * ref_mean).any())
            X.append(past_window)
            y.append(surge_flag)
    return np.array(X), np.array(y)

# Build dataset for surge prediction
X_surge, y_surge = make_surge_dataset(dataset2)

# Train Gradient Boosting Classifier
gbc = GradientBoostingClassifier(random_state=42)
gbc.fit(X_surge, y_surge)
y_pred = gbc.predict(X_surge)
y_prob = gbc.predict_proba(X_surge)[:,1]

print("Surge prediction accuracy:", accuracy_score(y_surge, y_pred))

# Confusion matrix visualization
cm = confusion_matrix(y_surge, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix (Surge Prediction)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# ROC curve visualization
fpr, tpr, _ = roc_curve(y_surge, y_prob)
roc_auc = auc(fpr, tpr)
plt.figure()
plt.plot(fpr, tpr, label=f"ROC curve (AUC = {roc_auc:.2f})")
plt.plot([0,1],[0,1],'--', color='gray')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Surge Prediction")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. If cases grow by 5% per day over 30 days
def project_cases_growth(initial_cases, growth_rate=0.05, days=30):
    cases = [initial_cases]
    for _ in range(days):
        cases.append(cases[-1] * (1 + growth_rate))
    return np.array(cases)

# Example: use the current number of confirmed cases for a country (e.g., US)
country = "US"
current_cases = dataset2[(dataset2["Country/Region"] == country)]["Confirmed"].iloc[-1]
projected_cases = project_cases_growth(current_cases)
print(f"Projected number of cases after 30 days at 5% daily growth: {int(projected_cases[-1]):,}")

plt.figure(figsize=(8,4))
plt.plot(projected_cases, label="Projected Cases (5%/day)")
plt.xlabel("Days")
plt.ylabel("Confirmed Cases")
plt.title(f"Projected COVID-19 Cases in {country} (5% Daily Growth)")
plt.legend()
plt.show()

# 2. If deaths decrease by 20% due to vaccination
current_deaths = dataset2[(dataset2["Country/Region"] == country)]["Deaths"].iloc[-1]
reduced_deaths = current_deaths * 0.8
print(f"Total deaths if reduced by 20% due to vaccination: {int(reduced_deaths):,}")

# 3. Simulate how the epidemic curve changes if 1 million vaccines are administered per week
def simulate_vaccine_effect(initial_cases, initial_vaccinated, weekly_vaccine, days=60, growth_rate=0.05, vaccine_efficacy=0.95):
    cases = [initial_cases]
    vaccinated = initial_vaccinated
    for day in range(1, days+1):
        if day % 7 == 0:
            vaccinated += weekly_vaccine
        # Assume vaccine efficacy reduces new cases proportionally to the protected population
        protected = vaccinated * vaccine_efficacy
        susceptible = max(0, 1 - protected / (current_cases + vaccinated))
        new_cases = cases[-1] * (1 + growth_rate * susceptible)
        cases.append(new_cases)
    return np.array(cases)

projected_with_vaccine = simulate_vaccine_effect(
    initial_cases=current_cases,
    initial_vaccinated=0,
    weekly_vaccine=1_000_000,
    days=60,
    growth_rate=0.05,
    vaccine_efficacy=0.95
)

plt.figure(figsize=(8,4))
plt.plot(projected_cases[:61], label="No Vaccine (5%/day)")
plt.plot(projected_with_vaccine, label="With 1M Vaccines per Week")
plt.xlabel("Days")
plt.ylabel("Confirmed Cases")
plt.title(f"Impact of Vaccination on COVID-19 Curve in {country}")
plt.legend()
plt.show()
```
